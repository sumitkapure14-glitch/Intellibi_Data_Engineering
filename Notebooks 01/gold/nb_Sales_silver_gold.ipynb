```python
# SCD Type: 1, UPSERT(Update else Insert), Merge Statement
```

```python
%sql
MERGE INTO sumit_catlog.sumit_gold01.dfMonthly_sales 
USING (
  SELECT DISTINCT * FROM sumit_catlog.sumit_silver01.dfMonthly_sales
) source
ON sumit_catlog.sumit_gold01.dfMonthly_sales.sale_id = source.sale_id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT * 
;
```
