```python
from pyspark.sql import functions as F
dfMonthlySalesClean = dfMonthly_sales.selectExpr(
    "CAST(sale_id AS string ) as sale_id",
    "UPPER(TRIM(product)) as product",
    "UPPER(TRIM(category)) as category",
    "CAST(quantity AS INT) as quantity",
    "CAST(price AS DOUBLE) as price",
    "CAST(sale_date AS DATE) as sale_date",
    "UPPER(TRIM(region)) as region"
)
dfMonthlySalesClean.display()
dfMonthlySalesClean.write \
    .mode("append") \
    .saveAsTable("sumit_catlog.sumit_bronze01.dfMonthly_sales")
# ['sale_id', 'product', 'category', 'quantity', 'price', 'sale_date', 'region']
```
